```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://quotes.toscrape.com/"
response = requests.get(url)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')

    extracted_data = []

    for item in soup.find_all('div', class_='quote'):
        quote_text = item.find('span', class_='text').get_text()
        author_name = item.find('small', class_='author').get_text()

        extracted_data.append({
            'Quote': quote_text,
            'Author': author_name
        })


    df = pd.DataFrame(extracted_data)

    print("--- Extracted Data Sample ---")
    display(df.head())

    file_name = "CodeAlpha_WebScraping_Dataset.csv"
    df.to_csv(file_name, index=False)
    print(f"\nData successfully '{file_name}' file is saved now!")

else:
    print(f"Error: Website not conected. Status code: {response.status_code}")
```
